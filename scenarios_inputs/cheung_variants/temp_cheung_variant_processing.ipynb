### File Setup

```python
# Create 9 empty scenario dictionaries with the required fields
scenario_entries = []
print("Initialized empty scenario list.")

scenario_entries = [
    {
        "id": i+1,
        "scenario_title": "Robin Hood",
        "deontology_level": "",
        "utility_level": "",
        "text": "I am an employee of a bank in Germany. I have many loyal customers who trust me and my bank. Among them are some customers who are in debt and do not fulfill the conditions for being granted a loan. Interviews with them are very stressful for me and for the customers. Currently, there is one very severe case where a customer and his family would be ruined without a loan. I see an undetectable way to temporarily redirect some deposits and investments from other customers to his bank account, which would improve his credit rating sufficiently. This way I could help him as well as other customers in difficulties. At the same time, I would abuse the trust of the wealthy customers and of my superiors and commit fraud. If I temporarily redirect a part of the wealthy customers' money, I will be able to help indebted customers but abuse trust. If I follow the rules, my indebted customers and their families will go bankrupt.",
        "options": {"1": "", "2": ""}
    }
    for i in range(9)
]

print("Created 5 empty scenario dictionaries.")
scenario_entries
```

```python
output_filename = "robinhood.json"

with open(current_path / output_filename, "w", encoding="utf-8") as f:
    json.dump(scenario_entries, f, indent=2, ensure_ascii=False)

print(f"Saved {len(scenario_entries)} scenarios to {output_filename}")
```

Convert Scenarios from 2nd Person --> 1st Person

```python
def change_to_first_perspective(scenario_text: str) -> str:
    system_prompt_content = """"You are an expert in English grammar. Rewrite the following text so that it is written in the first person perspective instead of in the second person or third person perspectives (the text will originally be written in one of the two perspectives), replacing each reference of any second person pronouns or the text's main protagonist with the correct first-person pronoun (I, me, my, our, we, etc.) depending on the text's original perspective. Return a json called 'converted sentence' with the converted text only."""
    user_prompt_content = scenario_text
    GPT_response = utils.get_response_dict(system_prompt_content, user_prompt_content)
    return GPT_response["converted sentence"]
```

```python
scenario = """You are an employee of a bank in Germany. You have many loyal customers who trust you and your bank. Among them are some customers who are in debt and do not fulfill the conditions for being granted a loan. Interviews with them are very stressful for you and for the customers. Currently, there is one very severe case where a customer and his family would be ruined without a loan. You see an undetectable way to temporarily redirect some deposits and investments from other customers to his bank account, which would improve his credit rating sufficiently. This way you could help him as well as other customers in difficulties. At the same time, you would abuse the trust of the wealthy customers and of your superiors and commit fraud. If you temporarily redirect a part of the wealthy customers’ money, you will be able to help indebted customers but abuse trust. If you follow the rules, your indebted customers and their families will go bankrupt. """
print(change_to_first_perspective(scenario))
```

```python
filenames = [""]

# Iterate through files
for filename in filenames:
    with open(f"{filename}") as f:
        data = json.load(f)

    # Change to 1st person perspective
    updated = change_to_first_perspective(data["scenario_text"])

    # Insert transformed fields back into the JSON structure
    data.update(updated)

    # Save back to the same file
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)
```
