## Analysis 3 -- Utility score vs. human moral judgment

**Question:** Are human moral-judgment ratings correlated with annotator utility scores at the scenario level?

The `min` + `non_i_only` pooling (worst-case impact on the patient/bystander) saturates at the floor for 41/48 scenarios, which could be masking a real relationship rather than ruling one out -- see `christensen_utility_funcs.DEFAULT_UTILITY_VARIANTS`. So instead of committing to one variant, this plots **all 6** `utility_score_<strategy>_<agg>` columns against human moral judgment side by side: 3 strategies (`non_i_only` = patient/bystander only, `i_only` = agent only, `simplest` = everyone pooled undifferentiated) x 2 aggregations (`min` = worst-case, `mean` = average-case). A rational, benefit-seeking agent would be more likely to commit the action when a given utility score is higher (less harmful); human moral_judgment is higher (more utilitarian) under the same logic, so the two should be directly comparable regardless of variant.

**Promising observation:** positive correlation, for at least one variant that isn't degenerate/saturated.

```python
# scenario_df is fine here (not scenario_df_rated) -- plot_scatter_reg drops NaN x/y rows internally,
# and this analysis has no second panel that needs a matching n.
utility_variant_cols = [c for c in scenario_df.columns if c.startswith("utility_score_")]
n = scenario_df["moral_judgment_mean"].notna().sum()
correlation_results = {}

for col in utility_variant_cols:
    variant_label = col.removeprefix("utility_score_")
    r, p = generic_analysis_utils.plot_scatter_reg(
        scenario_df,
        x=col,
        y="moral_judgment_mean",
        xlabel=f"Annotator Utility Score ({variant_label})",
        ylabel="Human Moral Judgment Rating (mean)",
        title=f"Utility Score ({variant_label}) vs. Human Moral Judgment",
    )
    correlation_results[variant_label] = (r, p)
    print(f"{variant_label}: Pearson r={r:.3f}, p={p:.4f}, n={n}")
```

```python
pd.DataFrame(
    [(name, r, p) for name, (r, p) in correlation_results.items()],
    columns=["utility_score_variant", "r", "p"],
).sort_values("r", ascending=False).reset_index(drop=True)
```

## Analysis 4 -- Do factor-level shifts in annotator utility mirror factor-level shifts in human ratings?

**Question:** For each of the 4 classification factors, does the annotator's utility score shift between levels the same way human moral-judgment ratings do?

Left panel = human moral judgment by factor level; right panel = annotator utility score by factor level. Each dot = one dilemma.

**Promising observation:** same ordering (which level is higher/lower) in both panels, for all 4 factors.

```python
import matplotlib.pyplot as plt

factors = ["evitability", "personal_force", "intentionality", "beneficial_to"]

for factor in factors:
    fig, axs = plt.subplots(1, 2, figsize=(14, 6))
    generic_analysis_utils.plot_bar_strip(
        scenario_df_rated, x=factor, y="moral_judgment_mean",
        ylabel="Human Moral Judgment (mean ± SEM)",
        title=f"Human Moral Judgment by {generic_analysis_utils.pretty_label(factor)}",
        ax=axs[0],
    )
    generic_analysis_utils.plot_bar_strip(
        scenario_df_rated, x=factor, y="utility_score",
        ylabel="Annotator Utility Score (mean ± SEM)",
        title=f"Annotator Utility Score by {generic_analysis_utils.pretty_label(factor)}",
        ax=axs[1],
    )
    fig.tight_layout()
    plt.show()
```

## Analysis 5 -- Do 2-way interaction effects show up similarly for annotator utility scores?

Same left/right (human vs. annotator) layout as Analysis 4, now with a `hue` for the second factor. Four hypothesized interactions (from Christensen et al.), each stated as a prediction about human willingness to commit harm:

1. **Personal Force x Beneficiary** -- if Personal, Self- vs Other-Beneficial should look similar; if Impersonal, Self-Beneficial should show higher willingness than Other-Beneficial.
2. **Personal Force x Intentionality** -- if Personal, Instrumental should show lower willingness than Accidental; if Impersonal, this difference should disappear.
3. **Intentionality x Beneficiary** -- if Instrumental, Other-Beneficial should show lower willingness than Self-Beneficial; if Accidental, Self-Beneficial should show lower willingness than Other-Beneficial (the pattern flips).
4. **Evitability x Intentionality** -- if Evitable, Instrumental should show lower willingness than Accidental; if Inevitable, this difference should disappear.

No single 'promising observation' here -- we're checking whether the annotator's utility score reproduces the same interaction pattern as the human ratings, whatever that pattern is.

```python
interaction_pairs = [
    ("personal_force", "beneficial_to"),
    ("personal_force", "intentionality"),
    ("intentionality", "beneficial_to"),
    ("evitability", "intentionality"),
]

for x_factor, hue_factor in interaction_pairs:
    fig, axs = plt.subplots(1, 2, figsize=(16, 6))
    generic_analysis_utils.plot_bar_strip(
        scenario_df_rated, x=x_factor, y="moral_judgment_mean", hue=hue_factor,
        ylabel="Human Moral Judgment (mean ± SEM)",
        title=f"Human Moral Judgment: {generic_analysis_utils.pretty_label(x_factor)} x {generic_analysis_utils.pretty_label(hue_factor)}",
        ax=axs[0],
    )
    generic_analysis_utils.plot_bar_strip(
        scenario_df_rated, x=x_factor, y="utility_score", hue=hue_factor,
        ylabel="Annotator Utility Score (mean ± SEM)",
        title=f"Annotator Utility Score: {generic_analysis_utils.pretty_label(x_factor)} x {generic_analysis_utils.pretty_label(hue_factor)}",
        ax=axs[1],
    )
    fig.tight_layout()
    plt.show()
```
